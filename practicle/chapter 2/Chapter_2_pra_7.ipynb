```python
import mysql.connector
```

```python
con = mysql.connector.connect(
    host="localhost",
    user="ashish",
    password="1234",
    database="studentdb"
)

print(con)
```

```python
cur = con.cursor()

cur.execute("create database if not exists studentdb")
cur.execute("use studentdb")
```

```python
cur.execute("""create table if not exists students(
id int primary key,
name varchar(50),
marks int
)""")

con.commit()
```

```python
sql = "insert into students values (%s,%s,%s)"
val = (1, "Ashish", 85)

cur.execute(sql, val)
con.commit()
```

```python
sql = "insert into students values (%s,%s,%s)"
val = [
    (2, "Rahul", 72),
    (3, "Jay", 90),
    (4, "Amit", 65)
]

cur.executemany(sql, val)
con.commit()
```

```python
cur.execute("select * from students")

data = cur.fetchall()

for i in data:
    print(i)
```

```python
cur.execute("select name, marks from students")

data = cur.fetchall()

for i in data:
    print(i)
```

```python
sql = "update students set marks = %s where id = %s"
val = (95, 2)

cur.execute(sql, val)
con.commit()
```

```python
cur.execute("select * from students")

for i in cur.fetchall():
    print(i)
```

```python
sql = "delete from students where id = %s"
val = (4,)

cur.execute(sql, val)
con.commit()
```
