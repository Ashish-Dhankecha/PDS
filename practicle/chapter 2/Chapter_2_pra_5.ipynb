```python
import csv

data = [
   ['Name', 'Age', 'City'],
   ['Alice', '30', 'New York'],
   ['Bob', '25', 'Los Angeles'],
   ['Charlie', '35', 'Chicago']
]
with open('abc2.csv', 'w') as c:
    writer = csv.writer(c)
    writer.writerows(data)
```

```python
with open('abc2.csv','r') as r:
    reader=csv.reader(r)
    for row in reader:
        print(row)
```
