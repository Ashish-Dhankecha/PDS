```python
import random as r 
import statistics as s
import math as m 
```

```python
print(r.randint(0,10))
```

```python
def xyz(a):
    d = 0
    for i in a:
        d+=i
    return d
def hello():
    print("hello")
def hellousr(a):
    print(f"hello {a}")
def sum():
    a = 20
    b = 30
    return a+b
def addmean(a):
    d = 0
    for i in a:
        d+=i
    return d,s.mean(a)
```

```python
c=[]
for i in range(10):
    c.append(r.randint(0,100))
print(c)
```

```python
d = xyz(c)
print(d)
hello()
hellousr("Ashish")
print(sum())
u , mean = addmean(c)
print(u)
print(mean)
```

```python
print(m.sqrt(d))      
print(m.factorial(c[0]))   
print(m.gcd(c[0], c[1]))    
print(m.lcm(c[0], c[1]))
```

```python
print("Mean:", s.mean(c)) 
print("Median:", s.median(c))  
print("Mode:", s.mode(c))  
print("Population stdev:", s.pstdev(c))
```
