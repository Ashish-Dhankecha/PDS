```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]
a = [5,15,25,27,12]
plt.plot(x, y, marker='o', color='blue', linewidth=2,label="FirstLine")
plt.plot(x, a, marker='o', linestyle='dashed',color='Red', linewidth=2,label="SecondLine")
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()

plt.show()
```
