# Pandas and Data Visualization

Small dataset for practicing data analysis and plots.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Student':['Riya','Aman','Neha','Karan','Mehul'],
    'Maths':[72, 84, 68, 91, 77],
    'Science':[75, 80, 70, 88, 82],
    'English':[69, 86, 74, 90, 79]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.shape)
print(df.columns)
print(df.head())
```

```python
print(df.describe())
```

```python
print(df['Maths'])
print(df[['Student','Science']])
```

```python
print(df.loc[1])
print(df.iloc[2:5])
```

```python
print(df[df['Maths'] > 75])
```

```python
df['Average'] = (df['Maths'] + df['Science'] + df['English']) / 3
print(df[['Student','Average']])
```

```python
print(df.sort_values('Average', ascending=False))
```

```python
print(df[['Maths','Science','English']].corr())
```

```python
top = df.loc[df['Average'].idxmax()]
print(top)
```

```python
plt.plot(df['Student'], df['Maths'], marker='o')
plt.title('Maths Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.show()
```

```python
plt.bar(df['Student'], df['Science'])
plt.title('Science Marks')
plt.show()
```

```python
plt.scatter(df['Maths'], df['English'])
plt.xlabel('Maths')
plt.ylabel('English')
plt.show()
```

```python
plt.hist(df['Science'], bins=4)
plt.title('Science Distribution')
plt.show()
```

```python
plt.boxplot(df['English'])
plt.title('English Box Plot')
plt.show()
```

```python
sns.barplot(x='Student', y='Average', data=df)
plt.title('Average Marks')
plt.show()
```

```python
sns.scatterplot(x='Maths', y='Science', data=df)
plt.show()
```

```python
sns.heatmap(df[['Maths','Science','English']].corr(), annot=True)
plt.show()
```

```python
sns.pairplot(df[['Maths','Science','English']])
plt.show()
```
