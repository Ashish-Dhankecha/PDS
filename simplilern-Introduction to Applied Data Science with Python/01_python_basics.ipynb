Some basic Python practice.

```python
def greet():
    print('Hello, Python')
```

```python
greet()
```

```python
def multiply(a, b):
    return a * b
```

```python
multiply(6, 7)
```

```python
x = 25
y = 4.5
z = -12
print(type(x))
print(type(y))
print(type(z))
```

```python
text = 'data science'
print(text)
print(len(text))
print(text[0:4])
print(text.upper())
```

```python
nums = [5, 10, 15, 20, 25]
print(nums)
print(nums[1])
print(nums[-2])
print(nums[1:4])
```

```python
nums.append(30)
print(nums)
nums.pop(2)
print(nums)
```

```python
for n in nums:
    print(n)
```

```python
student = {'name':'Rahul', 'age':20, 'marks':78}
print(student)
print(student['name'])
```

```python
student['grade'] = 'B'
print(student)
print(student.keys())
print(student.values())
```

```python
a = 12
b = 7
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
```

```python
a = 18
if a % 2 == 0:
    print('even')
else:
    print('odd')
```

```python
for i in range(1, 6):
    print(i * i)
```

```python
i = 5
while i > 0:
    print(i)
    i -= 1
```

```python
import math
print(math.sqrt(81))
print(math.factorial(5))
```
