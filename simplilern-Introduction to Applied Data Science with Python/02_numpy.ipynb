# NumPy Practice

Working with arrays and simple numerical operations.

```python
import numpy as np
```

```python
a = np.array([5, 10, 15, 20, 25])
print(a)
print(type(a))
```

```python
print(a[1])
print(a[-1])
print(a[1:4])
print(a.shape)
print(a.ndim)
```

```python
b = np.array([2, 4, 6, 8, 10])
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
print(np.sum(a))
print(np.mean(a))
print(np.median(a))
print(np.min(a))
print(np.max(a))
```

```python
x = np.arange(2, 22, 2)
print(x)
```

```python
print(np.zeros(6))
print(np.ones(4))
print(np.linspace(0, 1, 6))
```

```python
matrix = np.array([[2, 4, 6], [8, 10, 12], [14, 16, 18]])
print(matrix)
print(matrix.shape)
print(matrix[0])
print(matrix[:, 1])
```

```python
print(np.sum(matrix, axis=0))
print(np.sum(matrix, axis=1))
```

```python
print(np.amin(matrix, axis=0))
print(np.amax(matrix, axis=1))
```

```python
values = np.array([1, 2, 3, 4, 5, 6, 7, 8])
print(np.where(values % 2 == 0, 'even', 'odd'))
```

```python
result = np.select([values < 4, values >= 6], [values * 2, values + 10], default=values)
print(result)
```

```python
marks = np.array([42, 55, 63, 71, 88, 94])
print(np.percentile(marks, 50))
print(np.var(marks))
print(np.std(marks))
```

```python
angles = np.array([0, 30, 60, 90])
r = angles * np.pi / 180
print(np.sin(r))
print(np.cos(r))
```

```python
v = np.array([-2.3, 0.7, 4.9, -8.1])
print(np.floor(v))
print(np.ceil(v))
```

```python
p = np.array([[1, 2], [3, 4]])
q = np.array([[5, 6], [7, 8]])
print(np.dot(p, q))
print(p.T)
```
